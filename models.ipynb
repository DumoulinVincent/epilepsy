```python
import brian2 as b2
from neurodynex3.adex_model import AdEx
from neurodynex3.tools import plot_tools, input_factory

current = input_factory.get_step_current(10, 350, 1. * b2.ms, -65.0 * b2.pA)
state_monitor, spike_monitor = AdEx.simulate_AdEx_neuron(I_stim=current, simulation_time=500 * b2.ms,tau_m=1. * b2.ms,
                                                          R=500*b2.Mohm, v_rest=-70. * b2.mV, 
                                                          v_reset=-51. *b2.mV, v_rheobase=-50. * b2.mV, a=1.2 * b2.nS, b=7. * b2.pA,
                                                         v_spike=-30. * b2.mV, delta_T=11. * b2.mV, tau_w=100. * b2.ms)
plot_tools.plot_voltage_and_current_traces(state_monitor, current)
print("nr of spikes: {}".format(spike_monitor.count[0]))
```
