```python
from neuron  import h
from neuron.units import um
#import plotly
from matplotlib import pyplot, cm
h.load_file("stdlib.hoc")
h.load_file("import3d.hoc")

class Pyramidal:
    def _shift(self, x, y, z):
        soma=self.soma[0]
        n=soma.n3d()
        xs = [soma.x3d(i) for i in range(n)]
        ys = [soma.y3d(i) for i in range(n)]
        zs = [soma.z3d(i) for i in range(n)]
        ds = [soma.diam3d(i) for i in range(n)]
        for i, (a,b,c,d) in enumerate(zip(xs,ys,zs,ds)):
            soma.pt3dchange(i,a+x,b+y,c+z,d)
    def __init__(self, gid, x, y, z):
        self.gid = gid
        self.load_morphology()
        self._shift(x,y,z)
    def load_morphology(self):
        cell=h.Import3d_SWC_read()
        cell.input("c91662.swc")
        i3d=h.Import3d_GUI(cell, False)
        i3d.instantiate(self)


#pyr.apix, pyr.axon, pyr.soma, pyr.all etc

mypyrs = [Pyramidal(i,i*100,0,0) for i in range(10)]

ps=h.PlotShape(False)
ps.plot(pyplot)
pyplot.show()
```

```python
from neuron  import h, gui
from neuron.units import um, mV, ms, cm
import plotly
h.load_file("stdlib.hoc")
h.load_file("import3d.hoc")

class Pyramidal:
    def __init__(self, gid):
        self.gid = gid
        self.load_morphology()
        self.add_channels()
        self.discretize()
    def __repr__(self):
        return f"Pyramidal[{self.gid}]"
    def load_morphology(self):
        cell=h.Import3d_SWC_read()
        cell.input("c91662.swc")
        i3d=h.Import3d_GUI(cell, False)
        i3d.instantiate(self)
    def add_channels(self):
        h.hh.insert(self.all)
    def discretize(self):
        for sec in self.all:
            sec.nseg=1+2*int(sec.L/40)
pyr = Pyramidal(0)
```
