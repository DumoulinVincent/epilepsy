```python
%matplotlib inline
from netpyne import sim, specs
from neuron import h


netParams = specs.NetParams()  

PYRcell = {'secs': {}}
PYRcell['secs']['soma'] = {'geom': {}, 'mechs': {}}
PYRcell['secs']['soma']['geom'] = {'diam': 18.8, 'L': 18.8, 'Ra': 123.0}                           
PYRcell['secs']['soma']['mechs']['hh'] = {'gnabar': 0.12, 'gkbar': 0.036, 'gl': 0.003, 'el': -70}  
netParams.cellParams['PYR'] = PYRcell


netParams.popParams['S'] = {'cellType': 'PYR', 'numCells': 20}
netParams.popParams['M'] = {'cellType': 'PYR', 'numCells': 20}

netParams.synMechParams['AMPA'] = {'mod': 'Exp2Syn', 'tau1': 0.1, 'tau2': 5.0, 'e': 0}  

# netParams.stimSourceParams['bkg'] = {'type': 'NetStim', 'rate': 3, 'noise': 0.0}
# netParams.stimTargetParams['bkg->all'] = {
#     'source': 'bkg',
#     'conds': {'cellType': 'PYR'},
#     'weight': 0.01,
#     'delay': 5,
#     'synMech': 'AMPA'}
netParams.stimSourceParams['Stimu_S'] = {'type': 'NetStim', 'rate': 10, 'noise': 0.0}
netParams.stimTargetParams['Stimu_S->S'] = {
        'source': 'Stimu_S', 
        'conds': {'pop': 'S','cellList':list(range(15))}, 
        'weight': 0.01, 
        'delay': 5, 
        'synMech': 'AMPA'}
 

netParams.connParams['S->M'] = { 
        'preConds': {'pop': 'S'},   
        'postConds': {'pop': 'M'},  
        'probability': 0.5,     
        'weight': 0.01,             
        'delay': 5,                 
        'synMech': 'AMPA'} 
netParams.saveCellConns = True

simConfig = specs.SimConfig()

simConfig.duration = 100*1e3
simConfig.dt = 0.1
simConfig.verbose = False
simConfig.recordCellsSpikes = list(range(40))
simConfig.recordTraces = {'V_soma' : {'sec':'soma', 'loc' : 0.5, 'var' : 'v'}}
simConfig.recordStep = 0.1
simConfig.filename = 'Simu_HS'
simConfig.savePickle = False
simConfig.saveJson = False

simConfig.analysis['plotRaster'] = {'syncLines': True, 'saveFig': True}
simConfig.analysis['plotTraces'] = {'include': [1], 'saveFig': True}
simConfig.analysis['plotRateSpectrogram'] = {'include': {'all'}}
simConfig.analysis['plot2Dnet'] = {'saveFig': True}                   # plot 2D cell positions and connections
simConfig.analysis['plotSpikeHist'] = {'include': {'S','M'}, 'saveFig': True}


sim.create(netParams=netParams, simConfig=simConfig)
custom_conns = []

for pre_cell in sim.net.cells:
    if pre_cell.tags['pop'] != 'S':
        continue
    for post_cell in sim.net.cells:
        if post_cell.tags['pop'] != 'S' or pre_cell.gid == post_cell.gid:
            continue

        # Connect soma to soma
        syn = h.Exp2Syn(post_cell.secs['soma']['hObj'](0.5))
        syn.tau1 = 0.1
        syn.tau2 = 5.0
        syn.e = 0

        nc = h.NetCon(pre_cell.secs['soma']['hObj'](0.5)._ref_v, syn, sec=pre_cell.secs['soma']['hObj'])
        nc.weight[0] = 1e-6
        nc.delay = 5

        # Store metadata
        custom_conns.append({
            'preGid': pre_cell.gid,
            'postGid': post_cell.gid,
            'netcon': nc
        })

target_rate = 10 
grow_amount = 0.001  

from collections import Counter
def grow_connections(t):
    from collections import Counter

    spk_times = sim.simData['spkt']
    spk_ids = sim.simData['spkid']

    spike_counts = Counter()
    for spkid, spkt in zip(spk_ids, spk_times):
        if spkt <= t:
            spike_counts[spkid] += 1

    underactive = []
    active = []

    for cell in sim.net.cells:
        if cell.tags['pop'] == 'S':
            gid = cell.gid
            rate = spike_counts.get(gid, 0) / (t / 1000.0)
            print(f"  Cell {gid} firing rate: {rate:.2f} Hz")
            if rate < target_rate:
                underactive.append(gid)
            else:
                active.append(gid)

    for conn in custom_conns:
        if conn['preGid'] in active and conn['postGid'] in underactive:
            conn['netcon'].weight[0] += grow_amount
            print(f"Grew conn from {conn['preGid']} to {conn['postGid']} (new weight = {conn['netcon'].weight[0]:.4f})")

sim.runSimWithIntervalFunc(interval=1000, func=grow_connections)
sim.analyze()
#sim.scheduleTask(grow_schedule, dt=1000)
#sim.create()
#sim.runSim()
sim.analysis.plotRaster(syncLines= True)
```
